```python
# Cell 19: Install dependencies & detect real GPU
!pip install -q torch torchvision pynvml

import torch
import torchvision
import torchvision.transforms as transforms
import torch.nn as nn
import torch.optim as optim
from torch.cuda.amp import autocast, GradScaler
import subprocess
import threading

try:
    from pynvml import (nvmlInit, nvmlDeviceGetHandleByIndex,
                        nvmlDeviceGetUtilizationRates, nvmlDeviceGetMemoryInfo,
                        nvmlDeviceGetPowerUsage, nvmlDeviceGetTemperature)
    PYNVML_AVAILABLE = True
except ImportError:
    PYNVML_AVAILABLE = False
    print("pynvml not available, will use torch.cuda fallback for monitoring")

# Detect GPU
if not torch.cuda.is_available():
    raise RuntimeError("No GPU detected! Enable GPU in Kaggle/Colab settings.")

gpu_name = torch.cuda.get_device_name(0)
gpu_mem = torch.cuda.get_device_properties(0).total_memory / 1e9

# Map real GPU to pricing
GPU_PRICING = {"T4": 0.35, "A100": 3.67, "V100": 2.48, "P100": 1.46, "L4": 0.81}
detected_type = "T4"  # default
for gtype in GPU_PRICING:
    if gtype.lower() in gpu_name.lower():
        detected_type = gtype
        break

print(f"Real GPU Detected")
print(f"   Name:    {gpu_name}")
print(f"   Memory:  {gpu_mem:.1f} GB")
print(f"   Type:    {detected_type}")
print(f"   Pricing: ${GPU_PRICING[detected_type]:.2f}/hr (on-demand)")
print(f"   CUDA:    {torch.version.cuda}")
print(f"   pynvml:  {'available' if PYNVML_AVAILABLE else 'NOT available'}")
```

```python
# Cell 20: GPU Metrics Collection (works on Kaggle/Colab)
# Strategy: pynvml first (reliable Python lib), torch.cuda fallback

def get_gpu_metrics():
    """Get GPU metrics inline. Called during training loop."""

    # Method 1: pynvml (most reliable on Kaggle - pure Python, no subprocess)
    if PYNVML_AVAILABLE:
        try:
            nvmlInit()
            handle = nvmlDeviceGetHandleByIndex(0)
            util = nvmlDeviceGetUtilizationRates(handle)
            mem_info = nvmlDeviceGetMemoryInfo(handle)
            try:
                power = nvmlDeviceGetPowerUsage(handle) / 1000.0  # mW -> W
            except Exception:
                power = 0.0
            try:
                temp = nvmlDeviceGetTemperature(handle, 0)  # NVML_TEMPERATURE_GPU = 0
            except Exception:
                temp = 0.0
            return {
                'timestamp': time.time(),
                'gpu_util_pct': float(util.gpu),
                'mem_util_pct': float(util.memory),
                'mem_used_mb': mem_info.used / 1e6,
                'mem_total_mb': mem_info.total / 1e6,
                'power_watts': float(power),
                'temp_c': float(temp),
            }
        except Exception as e:
            pass  # Fall through to torch.cuda

    # Method 2: torch.cuda (always works, limited metrics)
    mem_used = torch.cuda.memory_allocated() / 1e6
    mem_reserved = torch.cuda.memory_reserved() / 1e6
    mem_total = torch.cuda.get_device_properties(0).total_memory / 1e6
    util_approx = (mem_reserved / mem_total) * 100 if mem_total > 0 else 0
    return {
        'timestamp': time.time(),
        'gpu_util_pct': util_approx,
        'mem_util_pct': util_approx,
        'mem_used_mb': mem_used,
        'mem_total_mb': mem_total,
        'power_watts': 0.0,
        'temp_c': 0.0,
    }

# === DIAGNOSTIC TEST ===
print("=" * 50)
print("GPU METRICS DIAGNOSTIC")
print("=" * 50)

# Test pynvml
print(f"\n1. pynvml available: {PYNVML_AVAILABLE}")
if PYNVML_AVAILABLE:
    try:
        nvmlInit()
        handle = nvmlDeviceGetHandleByIndex(0)
        util = nvmlDeviceGetUtilizationRates(handle)
        mem = nvmlDeviceGetMemoryInfo(handle)
        print(f"   pynvml works! GPU util={util.gpu}%, mem={mem.used/1e6:.0f}/{mem.total/1e6:.0f} MB")
        try:
            pwr = nvmlDeviceGetPowerUsage(handle) / 1000.0
            print(f"   Power: {pwr:.1f}W")
        except Exception as e:
            print(f"   Power: not available ({e})")
        try:
            tmp = nvmlDeviceGetTemperature(handle, 0)
            print(f"   Temp: {tmp}C")
        except Exception as e:
            print(f"   Temp: not available ({e})")
    except Exception as e:
        print(f"   pynvml FAILED: {e}")

# Test full function
print(f"\n2. get_gpu_metrics() test:")
test_result = get_gpu_metrics()
for k, v in sorted(test_result.items()):
    if k != 'timestamp':
        print(f"   {k}: {v}")

print(f"\n   Method: {'pynvml' if test_result.get('temp_c', 0) > 0 or (PYNVML_AVAILABLE and test_result.get('gpu_util_pct', -1) >= 0) else 'torch.cuda'}")
print(f"\nReady for training.")
```

```python
# Cell 21: Prepare CIFAR-10 dataset and ResNet-18 model
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform_train)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
]))
testloader = torch.utils.data.DataLoader(testset, batch_size=256, shuffle=False, num_workers=2)

print(f"Dataset ready: {len(trainset)} train / {len(testset)} test images")
print(f"Batches per epoch: {len(trainloader)}")


def create_model():
    """Create a fresh ResNet-18 for CIFAR-10."""
    model = torchvision.models.resnet18(weights=None, num_classes=10)
    model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
    model.maxpool = nn.Identity()
    return model.cuda()


def train_epoch_monitored(model, loader, optimizer, criterion, use_amp=False, sample_interval=10):
    """Train one epoch with inline GPU metrics collection every N batches."""
    model.train()
    total_loss, correct, total = 0, 0, 0
    scaler = GradScaler() if use_amp else None
    metrics_samples = []
    start = time.time()

    for batch_idx, (inputs, targets) in enumerate(loader):
        inputs, targets = inputs.cuda(), targets.cuda()
        optimizer.zero_grad()

        if use_amp:
            with autocast():
                outputs = model(inputs)
                loss = criterion(outputs, targets)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()

        total_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        correct += predicted.eq(targets).sum().item()
        total += inputs.size(0)

        # Collect GPU metrics every N batches (inline, no threading)
        if batch_idx % sample_interval == 0:
            torch.cuda.synchronize()
            sample = get_gpu_metrics()
            metrics_samples.append(sample)

    elapsed = time.time() - start
    return total_loss / total, correct / total * 100, elapsed, metrics_samples


print("Model & training functions ready (inline monitoring every 10 batches)")
```

```python
# Cell 22: Train FP32 (baseline) with inline GPU monitoring
NUM_EPOCHS = 3
price_per_hour = GPU_PRICING[detected_type]

print("=" * 60)
print("EXPERIMENT 1: FP32 Training (Baseline)")
print("=" * 60)

model_fp32 = create_model()
optimizer_fp32 = optim.SGD(model_fp32.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
criterion = nn.CrossEntropyLoss()

torch.cuda.synchronize()
torch.cuda.reset_peak_memory_stats()
fp32_start = time.time()

fp32_results = []
fp32_all_metrics = []
for epoch in range(NUM_EPOCHS):
    loss, acc, t, epoch_metrics = train_epoch_monitored(model_fp32, trainloader, optimizer_fp32, criterion, use_amp=False)
    fp32_results.append({"epoch": epoch+1, "loss": loss, "accuracy": acc, "time_s": t})
    fp32_all_metrics.extend(epoch_metrics)
    print(f"   Epoch {epoch+1}/{NUM_EPOCHS} | Loss: {loss:.4f} | Acc: {acc:.1f}% | Time: {t:.1f}s | Samples: {len(epoch_metrics)}")

fp32_total_time = time.time() - fp32_start
fp32_peak_mem = torch.cuda.max_memory_allocated() / 1e9
fp32_cost = (fp32_total_time / 3600) * price_per_hour

print(f"\n   Total samples collected: {len(fp32_all_metrics)}")

# Build monitor dataframe
fp32_monitor_df = pd.DataFrame(fp32_all_metrics)
if not fp32_monitor_df.empty and 'timestamp' in fp32_monitor_df.columns:
    fp32_monitor_df['elapsed'] = fp32_monitor_df['timestamp'] - fp32_monitor_df['timestamp'].iloc[0]

print(f"\n   FP32 Summary:")
print(f"   Total time:      {fp32_total_time:.1f}s")
print(f"   Peak memory:     {fp32_peak_mem:.2f} GB")
if not fp32_monitor_df.empty and 'gpu_util_pct' in fp32_monitor_df.columns:
    print(f"   Avg GPU util:    {fp32_monitor_df['gpu_util_pct'].mean():.1f}%")
    print(f"   Avg power:       {fp32_monitor_df['power_watts'].mean():.1f}W")
    print(f"   Avg temperature: {fp32_monitor_df['temp_c'].mean():.1f}C")
    print(f"   Max GPU util:    {fp32_monitor_df['gpu_util_pct'].max():.1f}%")
else:
    print(f"   GPU util:        DataFrame empty! columns={list(fp32_monitor_df.columns) if not fp32_monitor_df.empty else 'EMPTY'}")
    if fp32_all_metrics:
        print(f"   First sample keys: {list(fp32_all_metrics[0].keys())}")
print(f"   Estimated cost:  ${fp32_cost:.6f}")
```

```python
# Cell 23: Train Mixed Precision AMP (optimized) with inline GPU monitoring
print("=" * 60)
print("EXPERIMENT 2: Mixed Precision (AMP) Training")
print("=" * 60)

model_amp = create_model()
optimizer_amp = optim.SGD(model_amp.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)

torch.cuda.synchronize()
torch.cuda.reset_peak_memory_stats()
amp_start = time.time()

amp_results = []
amp_all_metrics = []
for epoch in range(NUM_EPOCHS):
    loss, acc, t, epoch_metrics = train_epoch_monitored(model_amp, trainloader, optimizer_amp, criterion, use_amp=True)
    amp_results.append({"epoch": epoch+1, "loss": loss, "accuracy": acc, "time_s": t})
    amp_all_metrics.extend(epoch_metrics)
    print(f"   Epoch {epoch+1}/{NUM_EPOCHS} | Loss: {loss:.4f} | Acc: {acc:.1f}% | Time: {t:.1f}s | Samples: {len(epoch_metrics)}")

amp_total_time = time.time() - amp_start
amp_peak_mem = torch.cuda.max_memory_allocated() / 1e9
amp_cost = (amp_total_time / 3600) * price_per_hour

print(f"\n   Total samples collected: {len(amp_all_metrics)}")

# Build monitor dataframe
amp_monitor_df = pd.DataFrame(amp_all_metrics)
if not amp_monitor_df.empty and 'timestamp' in amp_monitor_df.columns:
    amp_monitor_df['elapsed'] = amp_monitor_df['timestamp'] - amp_monitor_df['timestamp'].iloc[0]

print(f"\n   AMP Summary:")
print(f"   Total time:      {amp_total_time:.1f}s")
print(f"   Peak memory:     {amp_peak_mem:.2f} GB")
if not amp_monitor_df.empty and 'gpu_util_pct' in amp_monitor_df.columns:
    print(f"   Avg GPU util:    {amp_monitor_df['gpu_util_pct'].mean():.1f}%")
    print(f"   Avg power:       {amp_monitor_df['power_watts'].mean():.1f}W")
    print(f"   Avg temperature: {amp_monitor_df['temp_c'].mean():.1f}C")
    print(f"   Max GPU util:    {amp_monitor_df['gpu_util_pct'].max():.1f}%")
else:
    print(f"   GPU util:        DataFrame empty! columns={list(amp_monitor_df.columns) if not amp_monitor_df.empty else 'EMPTY'}")
    if amp_all_metrics:
        print(f"   First sample keys: {list(amp_all_metrics[0].keys())}")
print(f"   Estimated cost:  ${amp_cost:.6f}")
```

```python
# Cell 27: Multi-GPU Cost Analysis
import matplotlib.pyplot as plt

print("=" * 60)
print("EXERCISE 8.5.1: Multi-GPU Cost Analysis")
print("=" * 60)

def analyze_multi_gpu_cost(base_time_hours, gpu_type, gpu_counts, scaling_factors=None):
    """Analyze multi-GPU training costs with realistic sub-linear scaling.

    Returns dict with per-config metrics + the cost-per-performance optimum.
    """
    price = GPU_PRICING.get(gpu_type, 1.0)
    if scaling_factors is None:
        # Realistic sub-linear scaling (Amdahl-style): speedup = n ** 0.85
        scaling_factors = {n: 1.0 if n == 1 else (n ** 0.85) for n in gpu_counts}
    baseline_cost = base_time_hours * price
    rows = []
    for n in gpu_counts:
        eff_speedup = scaling_factors[n]
        time_h = base_time_hours / eff_speedup
        total_cost = time_h * price * n
        efficiency_pct = (eff_speedup / n) * 100
        cost_per_perf = total_cost / eff_speedup
        rows.append({
            "gpu_count": n,
            "speedup": eff_speedup,
            "time_hours": time_h,
            "total_cost": total_cost,
            "efficiency_pct": efficiency_pct,
            "cost_per_perf": cost_per_perf,
            "vs_baseline_pct": (total_cost / baseline_cost - 1) * 100,
        })
    optimal = min(rows, key=lambda r: r["cost_per_perf"])
    return {"gpu_type": gpu_type, "price": price, "rows": rows, "optimal": optimal}

base_training_time = 2.0  # hours for single GPU
test_gpu_counts = [1, 2, 4, 8]
multi_gpu_analysis = analyze_multi_gpu_cost(base_training_time, "A100", test_gpu_counts)

print(f"\nGPU type: {multi_gpu_analysis['gpu_type']} @ ${multi_gpu_analysis['price']:.2f}/hr")
print(f"Baseline: 1 GPU x {base_training_time}h = ${base_training_time*multi_gpu_analysis['price']:.2f}\n")
print(f"{'GPUs':>5} {'Speedup':>9} {'Time(h)':>9} {'Cost($)':>10} {'Eff%':>7} {'vs base':>9}")
print("-" * 56)
for r in multi_gpu_analysis["rows"]:
    print(f"{r['gpu_count']:>5} {r['speedup']:>8.2f}x {r['time_hours']:>9.2f} {r['total_cost']:>10.2f} {r['efficiency_pct']:>6.1f}% {r['vs_baseline_pct']:>+8.1f}%")

opt = multi_gpu_analysis["optimal"]
print(f"\n[OPTIMAL] {opt['gpu_count']} GPU(s) -- cost/perf ${opt['cost_per_perf']:.2f}, time {opt['time_hours']:.2f}h")
print("Insight: more GPUs reduce wall-clock time but add cost due to sub-linear scaling.")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
gpus = [r["gpu_count"] for r in multi_gpu_analysis["rows"]]
costs = [r["total_cost"] for r in multi_gpu_analysis["rows"]]
speedups = [r["speedup"] for r in multi_gpu_analysis["rows"]]
ax[0].plot(gpus, costs, "o-", color="#3b82f6", linewidth=2, markersize=8)
ax[0].set_xlabel("GPU count"); ax[0].set_ylabel("Total cost ($)")
ax[0].set_title("Total cost vs GPU count"); ax[0].grid(alpha=0.3)
ax[1].plot(gpus, speedups, "o-", color="#10b981", linewidth=2, label="actual")
ax[1].plot(gpus, gpus, "--", color="gray", label="linear (ideal)")
ax[1].set_xlabel("GPU count"); ax[1].set_ylabel("Speedup factor")
ax[1].set_title("Scaling efficiency"); ax[1].legend(); ax[1].grid(alpha=0.3)
plt.tight_layout()
plt.savefig("multi_gpu_scaling.png", dpi=120, bbox_inches="tight")
plt.show()
print("\n[SAVED] multi_gpu_scaling.png")
```

```python
# Cell 28: Project Cost Forecasting
import math
import matplotlib.pyplot as plt

print("=" * 60)
print("EXERCISE 8.5.2: Project Cost Forecasting")
print("=" * 60)

def forecast_project_cost(phases, contingency_pct=20, confidence_level=0.95):
    """Forecast multi-phase project cost with uncertainty + contingency.

    Uses a normal-approx confidence interval; z-score from confidence_level.
    """
    # Approximate z-score (95% -> 1.96, 90% -> 1.645, 99% -> 2.576)
    z_map = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
    z = z_map.get(round(confidence_level, 2), 1.96)

    phase_results = []
    base_cost = 0.0
    variance_sum = 0.0
    for ph in phases:
        price = GPU_PRICING.get(ph["gpu_type"], 1.0)
        ph_cost = ph["duration_hours"] * price * ph["gpu_count"]
        sigma = ph_cost * ph["uncertainty_pct"]  # std-dev estimate
        phase_results.append({
            "name": ph["name"],
            "gpu_type": ph["gpu_type"],
            "gpu_count": ph["gpu_count"],
            "duration_hours": ph["duration_hours"],
            "base_cost": ph_cost,
            "uncertainty_pct": ph["uncertainty_pct"],
            "sigma": sigma,
            "low": ph_cost - z * sigma,
            "high": ph_cost + z * sigma,
        })
        base_cost += ph_cost
        variance_sum += sigma ** 2

    total_sigma = math.sqrt(variance_sum)
    contingency = base_cost * (contingency_pct / 100.0)
    expected = base_cost + contingency
    ci_low = base_cost - z * total_sigma
    ci_high = base_cost + z * total_sigma + contingency

    return {
        "phases": phase_results,
        "base_cost": base_cost,
        "contingency": contingency,
        "expected_cost": expected,
        "confidence_level": confidence_level,
        "ci_low": max(0.0, ci_low),
        "ci_high": ci_high,
        "total_sigma": total_sigma,
    }

example_project = [
    {"name": "Data Preparation", "gpu_type": "T4", "gpu_count": 1, "duration_hours": 40, "uncertainty_pct": 0.15},
    {"name": "Model Training", "gpu_type": "A100", "gpu_count": 4, "duration_hours": 120, "uncertainty_pct": 0.25},
    {"name": "Hyperparameter Tuning", "gpu_type": "A100", "gpu_count": 8, "duration_hours": 60, "uncertainty_pct": 0.30},
    {"name": "Model Evaluation", "gpu_type": "T4", "gpu_count": 2, "duration_hours": 20, "uncertainty_pct": 0.10},
]

project_forecast = forecast_project_cost(example_project, contingency_pct=20, confidence_level=0.95)

print("\nPhase-by-phase forecast:")
print(f"{'Phase':<24} {'GPU':<6} {'#':>3} {'h':>5} {'Cost($)':>10} {'Low($)':>10} {'High($)':>10}")
print("-" * 78)
for p in project_forecast["phases"]:
    print(f"{p['name']:<24} {p['gpu_type']:<6} {p['gpu_count']:>3} {p['duration_hours']:>5.0f} "
          f"{p['base_cost']:>10.2f} {p['low']:>10.2f} {p['high']:>10.2f}")

print("\n" + "=" * 60)
print(f"Base cost (sum of phases):       ${project_forecast['base_cost']:>10.2f}")
print(f"Contingency (20% buffer):        ${project_forecast['contingency']:>10.2f}")
print(f"Expected cost (base+contingency):${project_forecast['expected_cost']:>10.2f}")
print(f"\n{int(project_forecast['confidence_level']*100)}% confidence interval:")
print(f"  Best case (low):               ${project_forecast['ci_low']:>10.2f}")
print(f"  Worst case (high):             ${project_forecast['ci_high']:>10.2f}")

# Plot
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
phase_names = [p["name"] for p in project_forecast["phases"]]
phase_costs = [p["base_cost"] for p in project_forecast["phases"]]
phase_lows = [p["low"] for p in project_forecast["phases"]]
phase_highs = [p["high"] for p in project_forecast["phases"]]
x = range(len(phase_names))
ax[0].bar(x, phase_costs, color="#3b82f6", alpha=0.85, label="Base cost")
err_low = [c - l for c, l in zip(phase_costs, phase_lows)]
err_high = [h - c for h, c in zip(phase_highs, phase_costs)]
ax[0].errorbar(x, phase_costs, yerr=[err_low, err_high], fmt="none", color="black", capsize=5)
ax[0].set_xticks(x); ax[0].set_xticklabels(phase_names, rotation=20, ha="right")
ax[0].set_ylabel("Cost ($)"); ax[0].set_title("Phase cost with 95% CI"); ax[0].grid(axis="y", alpha=0.3)

scenarios = ["Best case", "Expected", "Worst case"]
values = [project_forecast["ci_low"], project_forecast["expected_cost"], project_forecast["ci_high"]]
colors = ["#10b981", "#3b82f6", "#ef4444"]
ax[1].bar(scenarios, values, color=colors)
for i, v in enumerate(values):
    ax[1].text(i, v, f"${v:,.0f}", ha="center", va="bottom", fontweight="bold")
ax[1].set_ylabel("Total cost ($)"); ax[1].set_title("Project cost scenarios")
ax[1].grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("project_forecast.png", dpi=120, bbox_inches="tight")
plt.show()
print("\n[SAVED] project_forecast.png")
```

```python
# Cell 29: Optimization Opportunity Analysis
import matplotlib.pyplot as plt

print("=" * 60)
print("EXERCISE 8.5.3: Advanced Optimization Opportunity Analysis")
print("=" * 60)

EFFORT_SCORE = {"LOW": 1, "MEDIUM": 2, "HIGH": 3}
RISK_PENALTY = {"LOW": 1.0, "MEDIUM": 0.85, "HIGH": 0.65}

def analyze_optimization_opportunities(current_config, optimization_strategies):
    """Score and prioritize optimization strategies.

    priority = savings_pct * risk_penalty / effort_score (higher is better).
    Returns ranked list + cumulative savings if applied independently
    (multiplicative composition: remaining_cost *= (1 - s)).
    """
    price = GPU_PRICING.get(current_config["gpu_type"], 1.0)
    baseline_cost = current_config["duration_hours"] * price * current_config["gpu_count"]

    scored = []
    for s in optimization_strategies:
        effort = EFFORT_SCORE.get(s["implementation_effort"], 2)
        risk_pen = RISK_PENALTY.get(s["risk_level"], 0.85)
        priority = (s["savings_pct"] * risk_pen) / effort
        scored.append({
            **s,
            "abs_savings": baseline_cost * s["savings_pct"],
            "effort_score": effort,
            "risk_penalty": risk_pen,
            "priority": priority,
        })

    ranked = sorted(scored, key=lambda r: r["priority"], reverse=True)

    # Cumulative savings (multiplicative composition)
    remaining = 1.0
    cumulative_pct = []
    cumulative_abs = []
    for r in ranked:
        remaining *= (1 - r["savings_pct"])
        saved_pct = 1 - remaining
        cumulative_pct.append(saved_pct)
        cumulative_abs.append(baseline_cost * saved_pct)

    return {
        "baseline_cost": baseline_cost,
        "ranked": ranked,
        "cumulative_pct": cumulative_pct,
        "cumulative_abs": cumulative_abs,
    }

example_strategies = [
    {"name": "Switch to Mixed Precision (AMP)", "savings_pct": 0.25, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Use Spot Instances", "savings_pct": 0.60, "implementation_effort": "MEDIUM", "risk_level": "HIGH", "dependencies": []},
    {"name": "Optimize Batch Size", "savings_pct": 0.15, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Implement Early Stopping", "savings_pct": 0.20, "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
    {"name": "Switch to More Efficient GPU Type", "savings_pct": 0.40, "implementation_effort": "HIGH", "risk_level": "MEDIUM", "dependencies": []},
]

current_training_config = {
    "gpu_type": "A100",
    "gpu_count": 4,
    "duration_hours": 100,
    "precision": "FP32",
    "instance_type": "on-demand",
}

optimization_recommendations = analyze_optimization_opportunities(current_training_config, example_strategies)

print(f"\nBaseline cost: ${optimization_recommendations['baseline_cost']:.2f} "
      f"({current_training_config['gpu_count']}x {current_training_config['gpu_type']} "
      f"for {current_training_config['duration_hours']}h)\n")

print(f"{'Rank':>4} {'Strategy':<38} {'Save%':>6} {'Effort':>7} {'Risk':>7} {'Score':>7} {'$Save':>9}")
print("-" * 86)
for i, r in enumerate(optimization_recommendations["ranked"], start=1):
    print(f"{i:>4} {r['name']:<38} {r['savings_pct']*100:>5.0f}% {r['implementation_effort']:>7} {r['risk_level']:>7} {r['priority']:>7.3f} {r['abs_savings']:>9.0f}")

print("\nCumulative savings if applied in priority order:")
for i, r in enumerate(optimization_recommendations["ranked"]):
    cum_pct = optimization_recommendations["cumulative_pct"][i] * 100
    cum_abs = optimization_recommendations["cumulative_abs"][i]
    print(f"  After step {i+1} ({r['name']}): {cum_pct:>5.1f}% saved = ${cum_abs:,.2f}")

# Visualization: savings vs effort scatter + cumulative roadmap
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
for r in optimization_recommendations["ranked"]:
    color = {"LOW": "#10b981", "MEDIUM": "#f59e0b", "HIGH": "#ef4444"}[r["risk_level"]]
    ax[0].scatter(r["effort_score"], r["savings_pct"] * 100, s=200, color=color, alpha=0.8,
                  edgecolors="black", linewidth=1)
    ax[0].annotate(r["name"][:22], (r["effort_score"], r["savings_pct"] * 100),
                   xytext=(8, 5), textcoords="offset points", fontsize=9)
ax[0].set_xlabel("Implementation effort (1=Low, 3=High)")
ax[0].set_ylabel("Savings (%)")
ax[0].set_title("Optimization priority (color = risk)")
ax[0].set_xticks([1, 2, 3]); ax[0].set_xticklabels(["LOW", "MEDIUM", "HIGH"])
ax[0].grid(alpha=0.3)

steps = list(range(1, len(optimization_recommendations["ranked"]) + 1))
cum_abs = optimization_recommendations["cumulative_abs"]
ax[1].plot(steps, cum_abs, "o-", color="#3b82f6", linewidth=2, markersize=10)
ax[1].fill_between(steps, 0, cum_abs, alpha=0.2, color="#3b82f6")
ax[1].set_xlabel("Strategies applied (priority order)")
ax[1].set_ylabel("Cumulative savings ($)")
ax[1].set_title("Cumulative savings roadmap")
ax[1].set_xticks(steps); ax[1].grid(alpha=0.3)
for i, v in enumerate(cum_abs):
    ax[1].text(steps[i], v, f"${v:,.0f}", ha="center", va="bottom", fontsize=9)
plt.tight_layout()
plt.savefig("optimization_roadmap.png", dpi=120, bbox_inches="tight")
plt.show()
print("\n[SAVED] optimization_roadmap.png")
```

```python
# Cell 31: Challenge Exercise - Cost Optimization Strategy Design
print("=" * 60)
print("CHALLENGE EXERCISE 8.5.5: Design Cost Optimization Strategy")
print("=" * 60)

challenge_scenario = {
    "project": "Large Language Model Fine-tuning",
    "requirements": {
        "training_duration": 200,  # hours baseline
        "gpu_type": "A100",
        "gpu_count": 8,
        "precision": "FP32",
        "instance_type": "on-demand",
        "budget": 5000,
    },
    "constraints": {
        "max_preemption_risk": "MEDIUM",
        "min_accuracy": 0.95,
        "deadline": "2 weeks",
    }
}

req = challenge_scenario["requirements"]
con = challenge_scenario["constraints"]

print("\n[SCENARIO]")
print(f"  Project: {challenge_scenario['project']}")
print(f"  Baseline: {req['gpu_count']}x {req['gpu_type']} for {req['training_duration']}h ({req['precision']}, {req['instance_type']})")
print(f"  Budget:   ${con.get('budget', req.get('budget'))}    Deadline: {con['deadline']}    Max preemption risk: {con['max_preemption_risk']}")

# Step 1: Baseline cost
price = GPU_PRICING[req["gpu_type"]]
baseline_cost = req["training_duration"] * price * req["gpu_count"]
budget = req["budget"]
print(f"\n[STEP 1] BASELINE COST")
print(f"  {req['gpu_count']} x {req['gpu_type']} @ ${price:.2f}/hr x {req['training_duration']}h = ${baseline_cost:,.2f}")
print(f"  Budget: ${budget:,.2f}  ->  Over budget by ${baseline_cost - budget:,.2f} ({(baseline_cost/budget - 1)*100:.1f}%)")

# Step 2: Multi-GPU sweet spot
print(f"\n[STEP 2] MULTI-GPU ANALYSIS")
mg = analyze_multi_gpu_cost(req["training_duration"], req["gpu_type"], [1, 2, 4, 8, 16])
for r in mg["rows"]:
    print(f"  {r['gpu_count']:>2} GPU(s): time={r['time_hours']:.1f}h cost=${r['total_cost']:,.0f} eff={r['efficiency_pct']:.0f}%")
opt = mg["optimal"]
print(f"  -> Optimal (cost/perf): {opt['gpu_count']} GPU(s), time {opt['time_hours']:.1f}h, ${opt['total_cost']:,.0f}")

# Step 3: Select strategies (respect constraints: max_preemption_risk=MEDIUM)
print(f"\n[STEP 3] STRATEGY SELECTION (respecting max_preemption_risk={con['max_preemption_risk']})")
allowed_risks = {"LOW", "MEDIUM"}  # exclude HIGH because of constraint
selected_strategies = [
    {"name": "Switch to Mixed Precision (AMP)", "savings_pct": 0.25, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Optimize Batch Size", "savings_pct": 0.15, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Implement Early Stopping", "savings_pct": 0.20, "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
    {"name": "Switch to More Efficient GPU Type", "savings_pct": 0.40, "implementation_effort": "HIGH", "risk_level": "MEDIUM", "dependencies": []},
]
# We DROP "Use Spot Instances" since its risk=HIGH > MEDIUM constraint.
for s in selected_strategies:
    print(f"  + {s['name']:<38} save={s['savings_pct']*100:.0f}%  effort={s['implementation_effort']:<6} risk={s['risk_level']}")
print("  - Spot Instances DROPPED (risk=HIGH violates max_preemption_risk=MEDIUM)")

# Step 4: Cumulative savings on optimal GPU count
print(f"\n[STEP 4] CUMULATIVE SAVINGS (multiplicative composition)")
remaining_factor = 1.0
optimal_cost = opt["total_cost"]
for s in selected_strategies:
    remaining_factor *= (1 - s["savings_pct"])
final_cost_factor = remaining_factor
final_cost = optimal_cost * final_cost_factor
total_savings = 1 - final_cost_factor
print(f"  Cost factor after all strategies: {final_cost_factor:.3f} (= {total_savings*100:.1f}% saved)")
print(f"  Starting from optimal {opt['gpu_count']}-GPU cost ${optimal_cost:,.2f}")
print(f"  Final cost estimate: ${final_cost:,.2f}")

# Step 5: Forecast with contingency + uncertainty
print(f"\n[STEP 5] FORECAST WITH UNCERTAINTY (10% per strategy std-dev, 20% contingency)")
phase = [{
    "name": "Optimized LLM Fine-tuning",
    "gpu_type": req["gpu_type"],
    "gpu_count": opt["gpu_count"],
    "duration_hours": opt["time_hours"] * final_cost_factor,  # reduced by savings
    "uncertainty_pct": 0.20,
}]
forecast = forecast_project_cost(phase, contingency_pct=20, confidence_level=0.95)
print(f"  Base:        ${forecast['base_cost']:,.2f}")
print(f"  Contingency: ${forecast['contingency']:,.2f}")
print(f"  Expected:    ${forecast['expected_cost']:,.2f}")
print(f"  95% CI: [${forecast['ci_low']:,.2f}  ->  ${forecast['ci_high']:,.2f}]")

# Step 6: Verify budget + constraints
print(f"\n[STEP 6] VERIFICATION")
within_budget = forecast["expected_cost"] <= budget
worst_within = forecast["ci_high"] <= budget
print(f"  Expected (${forecast['expected_cost']:,.2f}) <= Budget (${budget:,.2f})?  {'YES' if within_budget else 'NO'}")
print(f"  Worst case (${forecast['ci_high']:,.2f}) <= Budget (${budget:,.2f})?      {'YES' if worst_within else 'NO'}")
print(f"  Preemption risk constraint (max {con['max_preemption_risk']}): SATISFIED (no HIGH-risk strategies)")
print(f"  Deadline ({con['deadline']}): wall-clock {phase[0]['duration_hours']:.1f}h on {opt['gpu_count']} GPU -> fits 2 weeks")

# Step 7: Strategy summary
print("\n" + "=" * 60)
print("[STRATEGY SUMMARY]")
print("=" * 60)
print(f"Baseline cost:              ${baseline_cost:,.2f}")
print(f"Optimized cost (expected):  ${forecast['expected_cost']:,.2f}")
print(f"Total saved:                ${baseline_cost - forecast['expected_cost']:,.2f}  ({(1 - forecast['expected_cost']/baseline_cost)*100:.1f}%)")
print(f"Within budget?              {'YES' if within_budget else 'NO'}")
print("\nRecommended actions (priority order):")
for i, s in enumerate(selected_strategies, 1):
    print(f"  {i}. {s['name']}  (~{s['savings_pct']*100:.0f}% savings, {s['implementation_effort']} effort)")
print(f"  {len(selected_strategies)+1}. Right-size cluster to {opt['gpu_count']} GPUs (optimal cost/perf)")
print("\nJustification:")
print("  * Multi-GPU sweet spot avoids diminishing returns of larger clusters")
print("  * AMP + batch tuning + early stopping = stacked savings with LOW risk")
print("  * Switching GPU type captures the largest single-axis savings (~40%)")
print("  * Spot rejected because risk_level=HIGH violates max_preemption_risk=MEDIUM")
print("  * Contingency buffer + 95% CI keeps worst-case projection observable for FinOps review")
```
